```python
import numpy as np
import matplotlib.pyplot as plt

```

```python
x_val = 1.0 #точка диффиреницования

# Формируем массив значений n от 1 до 21 и шагов h_n
n = np.arange(1, 22)
h = 2.0 / (2.0**n)


# Задаем исходные функции и их точные аналитические производные
def df1_exact(x): return 2 * x * np.cos(x**2)
def df2_exact(x): return -np.cos(x) * np.sin(np.sin(x))
def df3_exact(x): return -np.sin(x) * np.cos(np.cos(x)) * np.exp(np.sin(np.cos(x)))
def df4_exact(x): return 1.0 / (x + 3)
def df5_exact(x): return 0.5 * (x + 3)**(-0.5)

funcs = [
    {"name": r"$\sin(x^2)$", "f": lambda x: np.sin(x**2), "df": df1_exact},
    {"name": r"$\cos(\sin(x))$", "f": lambda x: np.cos(np.sin(x)), "df": df2_exact},
    {"name": r"$\exp(\sin(\cos(x)))$", "f": lambda x: np.exp(np.sin(np.cos(x))), "df": df3_exact},
    {"name": r"$\ln(x+3)$", "f": lambda x: np.log(x + 3), "df": df4_exact},
    {"name": r"$(x+3)^{0.5}$", "f": lambda x: (x + 3)**0.5, "df": df5_exact}
]


# Определяем 5 формул численного дифференцирования
def method_1(f, x, h):
    return (f(x + h) - f(x)) / h

def method_2(f, x, h):
    return (f(x) - f(x - h)) / h

def method_3(f, x, h):
    return (f(x + h) - f(x - h)) / (2 * h)

def method_4(f, x, h):
    term1 = (f(x + h) - f(x - h)) / (2 * h)
    term2 = (f(x + 2 * h) - f(x - 2 * h)) / (4 * h)
    return (4/3) * term1 - (1/3) * term2

def method_5(f, x, h):
    term1 = (f(x + h) - f(x - h)) / (2 * h)
    term2 = (f(x + 2 * h) - f(x - 2 * h)) / (4 * h)
    term3 = (f(x + 3 * h) - f(x - 3 * h)) / (6 * h)
    return (3/2) * term1 - (3/5) * term2 + (1/10) * term3

methods = [
    {"name": "Формула 1 (O(h))", "func": method_1},
    {"name": "Формула 2 (O(h))", "func": method_2},
    {"name": "Формула 3 (O(h^2))", "func": method_3},
    {"name": "Формула 4 (O(h^4))", "func": method_4},
    {"name": "Формула 5 (O(h^6))", "func": method_5}
]
```

```python

#Вычисление погрешностей и построение графиков
for item in funcs:
    plt.figure(figsize=(10, 6))
    
    exact_val = item["df"](x_val)
    

    log_h = np.log10(h)
    
    for m in methods:
        approx_val = m["func"](item["f"], x_val, h)
        error = np.abs(approx_val - exact_val)
        
        # Заменяем нули на машинный эпсилон, чтобы избежать ошибки при вычислении логарифма
        error = np.where(error == 0, np.finfo(float).eps, error)
        
        log_err = np.log10(error)
    

        label_str = m['name']

        
        plt.plot(log_h, log_err, marker='o', markersize=4, label=label_str)
        
    plt.title(f"Погрешность численного дифференцирования функции {item['name']} в точке $x = {x_val}$")
    
    plt.xlabel(r"Логарифм шага, $\lg h$")
    plt.ylabel(r"Логарифм абсолютной погрешности, $\lg \Delta$")

    plt.grid(True, which="both", linestyle=':', linewidth=1)
    plt.legend()
    
    plt.show()
```
